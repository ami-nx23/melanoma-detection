Extract the archive locally into `../data/melanoma_cancer_dataset/`. Keep the images out of Git unless redistribution is permitted by the dataset licence. The archive inspected for this project contains 9,605 training images and 1,000 test images.

## 4. Model training and evaluation
Run the full, leakage-aware model-selection workflow from a terminal:
```bash

```
This selects a model on a validation split from the training folder, refits it on the full training set, and evaluates the held-out test folder once. Review sensitivity, specificity, ROC-AUC, precision-recall and false negatives.

## 5. Limitations and responsible interpretation
The heuristic segmentation is not validated. Patient identifiers are unavailable, so patient-level splitting cannot be guaranteed. Dataset provenance/licence must be verified. Results do not establish clinical utility or generalisation to other populations or imaging devices.
